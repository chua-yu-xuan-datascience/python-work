##### $\textbf{1.2. Simple linear algebra in NumPy}$
**1.2.1. Dot product and matrix multiplication**
- `@` is matrix multiplication.
- `np.dot` / `x @ y` gives dot products (for 1D arrays).

```python
# Two 1D arrays (vectors)
x = np.array([1.0, 2.0, 3.0])
y = np.array([4.0, 5.0, 6.0])

# @ computes the dot product for 1D arrays
# This equals 1*4 + 2*5 + 3*6 = 32
dot = np.dot(x,y)
dot_xy = x @ y
dot
dot_xy
```

```python
M = np.array([[4.0, 4.0],
              [5.0, 6.0]])
v = np.array([8.0, 9.0])

# Matrix–vector multiplication
M @ v

# This computes:
# [1  2] [10]   [1*10 + 2*20]   [ 50]
# [3  4] [20] = [3*10 + 4*20] = [110]
```

**1.2.2. Solving a linear system**

Consider the follow system of equations:

\begin{align}
3x + 2y + z & = 4 \\
5x -5y + 4z & = 3 \\
6x +z & = 0
\end{align}

Then:

```python
# Let's solve the above system.
# Ax = b
# A = [3 2 1
    #  5 -5 4
    #  6 0 1] 
#x = [x
#     y
#     z]
#b = [4
#     3
#     0]

A    = np.array([[3, 2, 1],[5,-5,4],[6,0,1]])
b    = np.array([4,3,0])
sol  = np.linalg.solve(A,b) #to get vector of x 
print(sol)

# Other method (not preferred, above one is stable)
# Equivalently, we can pre-multiply the inverse of A to b:
sol1 = np.linalg.inv(A)@b             # requires that matrix A is nonsingular.
print(sol1)
```
